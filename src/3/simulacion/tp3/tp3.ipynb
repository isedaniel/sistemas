**(B)** Complete la función `u_initial` la cual genere un vector con la condición inicial dada:

```python
def u_initial (nx, L=4):
    """
    Vector con la condición inicial, dado un tamaño de grilla nx, para un valor máximo L=4
    """
    x = np.linspace(0, L, nx)
    u = np.ones(nx)
    u[x >= 2] = 0
    return u
```

**(C)** Escriba la función `computeF` la cual genere un vector $F(u)$.

```python
def computeF(u):
    """
    Generamos un vector que cumpla con F(u) = u**2 / 2
    """
    return u**2 / 2
```

**(D)** Empleando el esquema de MacCormack, complete la complete la función que se propone a continuación:

```python
def maccormack(u, nt, dt, dx):
    un = np.zeros((nt, len(u)))
    un[:] = u.copy()
    ustar = u.copy()
    
    for n in range(1, nt):
        F = computeF(u)
        ustar[:-1] = u[:-1] - dt / dx * (F[1:] - F[:-1])
        Fstar = computeF(ustar)
        un[n, 1:] = 0.5 * (u[1:] + ustar[1:] - dt / dx * (Fstar[1:] - Fstar[:-1]))
        u = un[n].copy()

    return un
```

```python
def maccormack(u, nt, dt, dx, eps=0.1):
    """Defino nuevamente maccormack con amortiguación. Agrego eps para manejar el valor epsilon."""
    un = np.zeros((nt, len(u)))
    un[:] = u.copy()
    ustar = u.copy()
    
    for n in range(1, nt):
        F = computeF(u)
        ustar[1:-1] = (u[1:-1] 
                       - dt / dx * (F[2:] - F[1:-1]) 
                       + eps * (u[2:] - 2 * u[1:-1] + u[:-2]))
        Fstar = computeF(ustar)
        un[n, 1:] = 0.5 * (u[1:] + ustar[1:] - dt / dx * (Fstar[1:] - Fstar[:-1]))
        u = un[n].copy()

    return un
```

## Trabajo de código: Tubo de choque

Definimos parámetros de simulación:

```python
nx = 81
dx = .25
dt = .0002
gamma = 1.4
```

Una nueva función para la condición inicial:

```python
def u_initial(nx, gamma=1.4):
    """
    Genera la matriz (3, nx) de variables conservativas iniciales [rho, rho*u, rho*eT]
    para el dominio x en [-10, 10].
    """
    x = np.linspace(-10, 10, nx)
    u = np.zeros((3, nx))
    
    # Condiciones primitivas iniciales (izquierda: x < 0, derecha: x >= 0)
    rho = np.where(x < 0, 1.0, 0.125)
    vel = np.zeros(nx)
    p = np.where(x < 0, 100e3, 10e3)  # En N/m^2 (Pa)
    
    # Llenado del vector de variables conservativas u = [rho, rho*u, rho*eT]
    u[0] = rho
    u[1] = rho * vel
    u[2] = p / (gamma - 1) + 0.5 * rho * vel**2
    
    return u
```

Defino función de flujo:

```python
def computeF(u, gamma=1.4):
    """
    Computa el vector de flujo F(u) = [rho*u, rho*u^2 + p, (rho*eT + p)*u]
    a partir del vector de variables conservativas u = [rho, rho*u, rho*eT].
    """
    rho = u[0]
    vel = u[1] / rho
    rho_eT = u[2]
    p = (gamma - 1) * (rho_eT - 0.5 * rho * vel**2)
    
    F = np.zeros_like(u)
    F[0] = u[1]
    F[1] = rho * vel**2 + p
    F[2] = (rho_eT + p) * vel
    
    return F
```

Defino la función para el método de Richtmyer:

```python
def richtmyer(u, nt, dt, dx):
    """
    Esquema de Richtmyer con u vector de tres variables
    """
    un = np.zeros((nt, u.shape[0], u.shape[1]))
    un[:] = u.copy()
    
    for n in range(1, nt):
        F = computeF(u)
        ustar = 0.5 * (u[:, 1:] + u[:, :-1]) - dt / (2 * dx) * (F[:, 1:] - F[:, :-1])
        Fstar = computeF(ustar)
        un[n, :, 1:-1] = u[:, 1:-1] - dt / dx * (Fstar[:, 1:] - Fstar[:, :-1])
        
        u = un[n].copy()
        
    return un
```

Para evaluar en $t=.01$, determino tamaño de `nt`:

```python
nt = int(.01 / dt) + 1
nt
```

Y opero esquema de Richtmyer:

```python
u = u_initial(nx, gamma)

un = richtmyer(u, nt, dt, dx)
```

Gráficamente:

```python
x = np.linspace(-10, 10, nx)
u_final = un[-1]

rho = u_final[0]
vel = u_final[1] / rho
p = (gamma - 1) * (u_final[2] - 0.5 * rho * vel**2)

fig, axs = plt.subplots(3, 1, figsize=(9, 10), sharex=True)

axs[0].plot(x, rho, color='#1f77b4', lw=2)
axs[0].set_ylabel('Densidad $\\rho$ [kg/m³]')
axs[0].set_title('Tubo de Choque (Método de Richtmyer) en $t = 0.01$ s')
axs[0].grid(True, linestyle='--', alpha=0.6)

axs[1].plot(x, vel, color='#2ca02c', lw=2)
axs[1].set_ylabel('Velocidad $u$ [m/s]')
axs[1].grid(True, linestyle='--', alpha=0.6)

axs[2].plot(x, p / 1000, color='#d62728', lw=2)
axs[2].set_ylabel('Presión $p$ [kN/m²]')
axs[2].set_xlabel('Posición $x$ [m]')
axs[2].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
